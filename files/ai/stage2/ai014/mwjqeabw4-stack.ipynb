# Reasons to implement stacks using lists/arrays:

Memory Efficient: Array elements do not hold the next elements address like linked list nodes do.

Easier to implement and understand: Using arrays to implement stacks require less code than using linked lists, and for this reason it is typically easier to understand as well.

A reason for not using arrays to implement stacks:

Fixed size: An array occupies a fixed part of the memory. This means that it could take up more memory than needed, or if the array fills up, it cannot hold more elements.



```python

# Stack Operations: PUSH, POP, TRAVERSE, UPDATE

MAX = 5
stack = []
top = -1


def push():
    global top

    if top == MAX - 1:
        print("Stack Overflow!")
    else:
        item = int(input("Enter the item to push: "))
        stack.append(item)
        top += 1
        print(item, "inserted successfully.")


def pop():
    global top

    if top == -1:
        print("Stack Underflow! Stack is empty.")
    else:
        item = stack.pop()
        top -= 1
        print("Deleted item:", item)


def traverse():
    if top == -1:
        print("Stack is empty.")
    else:
        print("Stack elements (TOP to BOTTOM):")
        for i in range(top, -1, -1):
            print(stack[i])


def update():
    if top == -1:
        print("Stack is empty.")
    else:
        pos = int(input("Enter the position to update (from TOP): "))

        if pos < 1 or pos > top + 1:
            print("Invalid position!")
        else:
            item = int(input("Enter the new value: "))
            index = top - pos + 1
            old_item = stack[index]
            stack[index] = item
            print(old_item, "updated to", item, "successfully.")


# Main Menu
while True:
    print("\n===== STACK OPERATIONS =====")
    print("1. PUSH")
    print("2. POP")
    print("3. TRAVERSE")
    print("4. UPDATE")
    print("5. EXIT")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        push()

    elif choice == 2:
        pop()

    elif choice == 3:
        traverse()

    elif choice == 4:
        update()

    elif choice == 5:
        print("Program terminated.")
        break

    else:
        print("Invalid choice! Please try again.")
```

PUSH	O(1)

TRAVERSE	O(n)
UPDATE	O(1)

global	Access and modify a global variable inside a function.
local	A variable defined inside a function.
top	Tracks the position of the last element in the stack.

Peek: Returns the top (last) element on the stack.
isEmpty: Checks if the stack is empty.
Size: Finds the number of elements in the stack.

```python

# Stack Operations: PEEK, isEmpty, SIZE

MAX = 5
stack = []
top = -1


# PEEK Operation
def peek():
    if top == -1:
        print("Stack is Empty!")
    else:
        print("Top element:", stack[top])


# isEmpty Operation
def isEmpty():
    if top == -1:
        print("Stack is Empty.")
    else:
        print("Stack is Not Empty.")


# SIZE Operation
def size():
    print("Number of elements:", top + 1)


# Main Menu
while True:
    print("\n===== STACK OPERATIONS =====")
    print("1. PEEK")
    print("2. isEmpty")
    print("3. SIZE")
    print("4. EXIT")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        peek()

    elif choice == 2:
        isEmpty()

    elif choice == 3:
        size()

    elif choice == 4:
        print("Program terminated.")
        break

    else:
        print("Invalid choice! Please try again.")
```
