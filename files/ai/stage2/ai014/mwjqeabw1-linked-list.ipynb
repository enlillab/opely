# 1- traverse

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

def traverseAndPrint(head):
  currentNode = head
  while currentNode:
    print(currentNode.data, end=" -> ")
    currentNode = currentNode.next
  print("null")

node1 = Node(7)
node2 = Node(11)
node3 = Node(3)
node4 = Node(2)
node5 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

traverseAndPrint(node1)
```

# 2- Delete a Node in a Linked List

If you want to delete a node in a linked list, it is important to connect the nodes on each side of the node before deleting it, so that the linked list is not broken.

So before deleting the node, we need to get the next pointer from the previous node, and connect the previous node to the new next node before deleting the node in between.

Also, it is a good idea to first connect next pointer to the node after the node we want to delete, before we delete it. This is to avoid a 'dangling' pointer, a pointer that points to nothing, even if it is just for a brief moment.

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

def traverseAndPrint(head):
  currentNode = head
  while currentNode:
    print(currentNode.data, end=" -> ")
    currentNode = currentNode.next
  print("null")

def deleteSpecificNode(head, nodeToDelete):
  if head == nodeToDelete:
    return head.next

  currentNode = head
  while currentNode.next and currentNode.next != nodeToDelete:
    currentNode = currentNode.next

  if currentNode.next is None:
    return head

  currentNode.next = currentNode.next.next

  return head

node1 = Node(7)
node2 = Node(11)
node3 = Node(3)
node4 = Node(2)
node5 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

print("Before deletion:")
traverseAndPrint(node1)

# Delete node4
node1 = deleteSpecificNode(node1, node4)

print("\nAfter deletion:")
traverseAndPrint(node1)
```

# 3- Insert a Node in a Linked List

class Node: إنشاء صنف اسمه Node.



__init__: دالة تُنفّذ تلقائيًا عند إنشاء عقدة جديدة.



self.data = data: تخزين القيمة داخل العقدة.



self.next = None: جعل الرابط فارغًا في البداية؛ لأن العقدة لم تُربط بأي عقدة أخرى بعد.

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

def traverseAndPrint(head):
  currentNode = head
  while currentNode:
    print(currentNode.data, end=" -> ")
    currentNode = currentNode.next
  print("null")

def insertNodeAtPosition(head, newNode, position):
  if position == 1:
    newNode.next = head
    return newNode

  currentNode = head
  for _ in range(position - 2):
    if currentNode.next is None:
      break
    currentNode = currentNode.next

  newNode.next = currentNode.next
  currentNode.next = newNode
  return head

node1 = Node(7)
node2 = Node(3)
node3 = Node(2)
node4 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4

print("Original list:")
traverseAndPrint(node1)

# Insert a new node with value 97 at position 2
newNode = Node(97)
node1 = insertNodeAtPosition(node1, newNode, 2)

print("\nAfter insertion:")
traverseAndPrint(node1)
```

currentNode = head: نبدأ من أول عقدة في القائمة.



while currentNode: نستمر طالما أن العقدة الحالية موجودة وليست None.



print(currentNode.data, end=" -> "): نطبع قيمة العقدة، ثم سهمًا دون الانتقال إلى سطر جديد.



currentNode = currentNode.next: ننتقل إلى العقدة التالية.



print("null"): عندما نصل إلى نهاية القائمة، نطبع null.


هنا نستخدم المتغير currentNode للوصول إلى العقدة التي تسبق الموقع المطلوب مباشرة.



لماذا نستخدم position - 2؟

لأننا نريد الوقوف عند العقدة السابقة للعقدة الجديدة، وليس عند الموقع المطلوب نفسه.

مثلًا، عند الإدراج في الموقع الثاني:

position - 2 = 2 - 2 = 0


لذلك لن تُنفَّذ الحلقة حتى مرة واحدة، وسيبقى currentNode عند العقدة الأولى، التي تحتوي على 7.

أما إذا أردنا الإدراج في الموقع الثالث:

position - 2 = 3 - 2 = 1


فتُنفَّذ الحلقة مرة واحدة، وينتقل currentNode من العقدة الأولى إلى العقدة الثانية.

ما معنى for _ in range(...)؟

الشرطة السفلية _ هنا اسم متغير لا نحتاج إلى استخدام قيمته؛ فنحن نهتم بعدد مرات تكرار الحلقة فقط.

# 4- Searching

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

def traverseAndPrint(head):
  currentNode = head
  while currentNode:
    print(currentNode.data, end=" -> ")
    currentNode = currentNode.next
  print("null")

def searchNode(head, targetValue):
  currentNode = head
  position = 1
  
  while currentNode:
    if currentNode.data == targetValue:
      return position
    currentNode = currentNode.next
    position += 1
    
  return -1

# Create a linked list
node1 = Node(7)
node2 = Node(3)
node3 = Node(2)
node4 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4

print("Linked List:")
traverseAndPrint(node1)

# Search for value 2
target = 2
pos = searchNode(node1, target)
if pos != -1:
  print(f"\nValue {target} found at position {pos}.")
else:
  print(f"\nValue {target} not found in the list.")

# Search for value 100
target = 100
pos = searchNode(node1, target)
if pos != -1:
  print(f"Value {target} found at position {pos}.")
else:
  print(f"Value {target} not found in the list.")
```

# 5- Sorting

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

def traverseAndPrint(head):
  currentNode = head
  while currentNode:
    print(currentNode.data, end=" -> ")
    currentNode = currentNode.next
  print("null")

def sortLinkedList(head):
  if not head:
    return head
#return none  #Empty list


  # Bubble Sort implementation for Linked List
  end = None #variable call end
  while end != head.next: #تتحكم هذه الحلقة في تكرار عمليات الترتيب
    p = head
    while p.next != end:  
      q = p.next
      if p.data > q.data:
        p.data, q.data = q.data, p.data
      p = p.next
    end = p #بعد انتهاء الحلقة الداخلية، يشير p إلى آخر عقدة تم الوصول إليها في هذا المرور
  return head

# Create an unsorted linked list
node1 = Node(7)
node2 = Node(3)
node3 = Node(97)
node4 = Node(2)
node5 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

print("Original list:")
traverseAndPrint(node1)

# Sort the linked list
node1 = sortLinkedList(node1)

print("\nSorted list:")
traverseAndPrint(node1)
```

def sortLinkedList(head): تعريف دالة تستقبل head، وهو مرجع إلى أول عقدة في القائمة.
if not head: يتحقق مما إذا كانت القائمة فارغة، أي إن head يساوي None.
return head يعيد None إذا كانت القائمة فارغة، وينهي الدالة دون محاولة ترتيب عناصر غير موجودة.
هذا الشرط مهم لتجنب محاولة الوصول إلى عقدة غير موجودة.


# 6- Updating

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

def traverseAndPrint(head):
  currentNode = head
  while currentNode:
    print(currentNode.data, end=" -> ")
    currentNode = currentNode.next
  print("null")

def updateNodeAtPosition(head, position, newValue):
  if not head:
    print("List is empty.")
    return head

  currentNode = head
  currentPosition = 1

  while currentNode and currentPosition < position:
    currentNode = currentNode.next
    currentPosition += 1

  if currentNode is None:
    print(f"Position {position} is out of bounds.")
  else:
    currentNode.data = newValue

  return head

# Create a linked list
node1 = Node(7)
node2 = Node(3)
node3 = Node(2)
node4 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4

print("Original list:")
traverseAndPrint(node1)

# Update the value at position 3 to 99
node1 = updateNodeAtPosition(node1, 3, 99)

print("\nAfter update operation (changing position 3 to 99):")
traverseAndPrint(node1)
```

# 7- Merge

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

def traverseAndPrint(head):
  currentNode = head
  while currentNode:
    print(currentNode.data, end=" -> ")
    currentNode = currentNode.next
  print("null")

def mergeSortedLists(head1, head2):
  # Create a dummy node to act as the start of the merged list
  dummy = Node(0)
  tail = dummy

  while True:
    if head1 is None:
      tail.next = head2
      break
    if head2 is None:
      tail.next = head1
      break

    if head1.data <= head2.data:
      tail.next = head1
      head1 = head1.next
    else:
      tail.next = head2
      head2 = head2.next
    
    tail = tail.next

  return dummy.next

# Create first sorted list: 1 -> 3 -> 5
list1_node1 = Node(1)
list1_node2 = Node(3)
list1_node3 = Node(5)
list1_node1.next = list1_node2
list1_node2.next = list1_node3

# Create second sorted list: 2 -> 4 -> 6
list2_node1 = Node(2)
list2_node2 = Node(4)
list2_node3 = Node(6)
list2_node1.next = list2_node2
list2_node2.next = list2_node3

print("First Sorted List:")
traverseAndPrint(list1_node1)

print("Second Sorted List:")
traverseAndPrint(list2_node1)

# Merge the lists
merged_head = mergeSortedLists(list1_node1, list2_node1)

print("\nMerged Sorted List:")
traverseAndPrint(merged_head)
```
